```python
"""
Buy:-

1 < 1000 = 0 discount
1000 >= 5000 = 10% discount
5000 + = 20% discount

"""
```

```python
Buy_amount = int(input("Enter your Buying Amount : "))
```

```python
if Buy_amount < 1000:
    print("NOT DISCOUNT")
elif Buy_amount >= 1000 and Buy_amount <= 5000:
    print("Actual Amount :", Buy_amount)
    discount_10 = Buy_amount * 10 / 100
    print("Discount Amount :", discount_10) 
    Actual_amount_10 = Buy_amount - discount_10
    print("Total Amount :",Actual_amount_10)  
else:
    print("Actual_amount :", Buy_amount)
    discount_20 = Buy_amount * 20 / 100
    print("Discount Amount :",discount_20) 
    Actual_amount_20 = Buy_amount - discount_20
    print("Total Amount :",Actual_amount_20)
```

```python
import random

bill = random.randint(1, 100)
print("Your electrical bill :",bill)
```

```python
def E_bill(pay):
    if bill == pay:
        return "Your Payment is successfull.."
    elif bill > pay :
         return f"your Payment is successfull but '{bill - pay} Rupess' is still pending.."
    else :
        return f"Your Payment is successfull but '{pay - bill} Rupess' is extra.."  
```

```python
pay = int(input("Pay your electrical bill :"))
print(E_bill(pay))
```
