```python
# Extract all numbers greater than 20 from [12, 25, 8, 40, 15, 50].

arr13 = np.array([12, 25, 8, 40, 15, 50])
arr14 = []

for i in arr13:
    if i > 20:
        arr14.append(i)
print(arr14)        
```
